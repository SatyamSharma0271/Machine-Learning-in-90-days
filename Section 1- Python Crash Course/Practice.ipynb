#### List comprehension

syntax = [< what you want to return> < iteration in  iterable> < condition to filter the elements>]

example:
> if you want to return squares of all the elements of a list where the list is : 
> lst  = [1,2,3,4,5]

> then you can write: [i*i for i in lst]

> if you want to return squares of all the even elements of a list where the list is : 
> lst  = [1,2,3,4,5]

> then you can write: [i*i for i in lst if i%2 == 0]
